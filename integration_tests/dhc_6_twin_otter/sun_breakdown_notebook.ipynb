```python
from fastga_he.gui.lcc_cost import (
    lcc_production_cost_bar_chart_simple,
    lcc_operation_cost_bar_chart_simple,
)

fig = lcc_production_cost_bar_chart_simple(
    [BASELINE_RESULT, HYBRID_RESULT, H2_RESULT],
    names_aircraft=["DHC6 Twin Otter", "Hybrid Retrofit DHC6", "Hydrogen Retrofit DHC6"],
    color_list=["#F59E0B", "#5A8F5A", "#6CC5F0"],
    item_filter_list=[
        "MSP",
        "Manufacturing",
        "Certification",
        "Engineering",
        "Powertrain",
        "Tooling",
        "Quality control",
    ],
)
fig.update_layout(
    title_text="Production Cost Share per Unit",
    title_font=dict(size=32),
    legend_font=dict(size=32),
    xaxis=dict(
        ticks="outside",
        title_font=dict(size=32),
        tickfont=dict(size=32),
        showline=True,
        linecolor="black",
        linewidth=3,
    ),
    yaxis=dict(
        ticks="outside",
        showline=True,
        linecolor="black",
        gridcolor="lightgrey",
        linewidth=3,
        tickfont=dict(size=32),
        title=dict(text="Cost [USD/unit]", font=dict(size=32)),
    ),
)
fig.update_layout(legend=dict(orientation="h", yanchor="top", y=1.02, xanchor="center", x=0.5))
# fig.update_layout(font=dict(size=16))
fig.write_html("chart_production.html")
```

```python
fig = lcc_operation_cost_bar_chart_simple(
    [BASELINE_RESULT, HYBRID_RESULT, H2_RESULT],
    names_aircraft=["DHC6 Twin Otter", "Hybrid Retrofit DHC6", "Hydrogen Retrofit DHC6"],
    item_filter_list=[
        "maintenance_cost",
        "annual_overhaul_deposit",
        "miscellaneous_cost",
        "annual_fuel_cost",
    ],
    color_list=["#F59E0B", "#6CC5F0", "#5A8F5A"],
)
fig.update_layout(
    title_text="Annual Operation Cost Comparison",
    title_font=dict(size=32),
    legend_font=dict(size=32),
    xaxis=dict(
        ticks="outside",
        title_font=dict(size=32),
        tickfont=dict(size=32),
        showline=True,
        linecolor="black",
        linewidth=3,
    ),
    yaxis=dict(
        ticks="outside",
        showline=True,
        linecolor="black",
        gridcolor="lightgrey",
        linewidth=3,
        tickfont=dict(size=32),
        title=dict(text="Cost [USD/year]", font=dict(size=32)),
    ),
)
fig.update_layout(legend=dict(orientation="h", yanchor="top", y=1.02, xanchor="center", x=0.5))
fig.write_html("bar_operation_chart.html")
```
